```python
import pandas as pd
import matplotlib.pyplot as plt
# from .utils import show_graph_model
```

```python
accuracy = pd.read_csv('../docs/results/accuracy.csv')

# Filter to only include rows with "(recorded data)" in their label
accuracy = accuracy[~accuracy['label'].str.contains('(recorded data)')]

print(len(accuracy))
```

```python
def show_charts(chart_df, sequence_length):
    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylim(0, 100)
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title('Accuracy of models with sequence length ' + str(sequence_length))
    plt.show()

    # Find accuracy min
    min_accuracy = chart_df['accuracy'].min()

    # Show accuracy in bar chart with min of 80 (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylim(min_accuracy - 2, 100)
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title(f'Accuracy of models with sequence length {str(sequence_length)} (min {int(min_accuracy - 2)}%)')
    plt.show()

    # Show training time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['training_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Training time (s)')
    plt.xlabel('Model')
    plt.title('Training time of models with sequence length ' + str(sequence_length))
    plt.show()

    # Show response time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['avg_response_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Response time (ms)')
    plt.xlabel('Model')
    plt.title('Response time of models with sequence length ' + str(sequence_length))
    plt.show()
```

```python
def show_accuracy_sequence_length(chart_df, model_name):
    # Create new column where each unique sequence length + sample_rate is a label
    chart_df = chart_df.sort_values(by='sequence_length')
    chart_df['chart_label'] = chart_df['sequence_length'].astype(int).astype(str)

    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['chart_label'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])

    # Set minimum to min value of accuracy - 2
    plt.ylim(bottom=50, top=100)

    plt.ylabel('Accuracy (%)')
    plt.xlabel('Sequence length')
    plt.xticks(rotation=45)
    plt.title(f'Accuracy of {model_name} model per sequence length')
    plt.show()

def show_accuracy_sampling_rate(chart_df, model_name):
    # Create new column where each unique sequence length + sample_rate is a label
    chart_df = chart_df.sort_values(by='sequence_length')
    chart_df['chart_label'] = chart_df['sampling_rate'].astype(str)

    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['chart_label'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])

    # Set minimum to min value of accuracy - 2
    plt.ylim(bottom=50, top=100)

    plt.ylabel('Accuracy (%)')
    plt.xlabel('Sampling rate divisor')
    plt.xticks(rotation=45)
    plt.title(f'Accuracy of {model_name} model per sampling rate divisor')
    plt.show()
```

```python
def show_results_model(model):
    model_df = accuracy[accuracy['model'] == model]

    show_accuracy_sequence_length(model_df[model_df['sampling_rate'] == 1], f'{model} full sampling rate')
    show_accuracy_sequence_length(model_df[model_df['sampling_rate'] == 2], f'{model} halved sampling rate')
    show_accuracy_sequence_length(model_df[model_df['sampling_rate'] == 4], f'{model} quarter sampling rate')

    model_256_df = model_df[model_df['sequence_length'] == 256]
    show_accuracy_sampling_rate(model_256_df, f'{model} with 256 samples per sequence')

    model_64_df = model_df[model_df['sequence_length'] == 64]
    show_accuracy_sampling_rate(model_64_df, f'{model} with 64 samples per sequence')

    model_128_df = model_df[model_df['sequence_length'] == 32]
    show_accuracy_sampling_rate(model_128_df, f'{model} with 32 samples per sequence')
```
