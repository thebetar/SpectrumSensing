```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
accuracy = pd.read_csv('../docs/results/accuracy.csv')
```

```python
def show_charts(chart_df, sequence_length):
    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title('Accuracy of models with sequence length ' + str(sequence_length))
    plt.show()

    # Find accuracy min
    min_accuracy = chart_df['accuracy'].min()

    # Show accuracy in bar chart with min of 80 (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylim(min_accuracy - 2, 100)
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title(f'Accuracy of models with sequence length {str(sequence_length)} (min {int(min_accuracy - 2)}%)')
    plt.show()

    # Show training time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['training_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Training time (s)')
    plt.xlabel('Model')
    plt.title('Training time of models with sequence length ' + str(sequence_length))
    plt.show()

    # Show response time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['avg_response_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Response time (ms)')
    plt.xlabel('Model')
    plt.title('Response time of models with sequence length ' + str(sequence_length))
    plt.show()
```

```python
accuracy_256 = accuracy[accuracy['sequence_length'] == 256]
accuracy_128 = accuracy[accuracy['sequence_length'] == 128]
accuracy_64 = accuracy[accuracy['sequence_length'] == 64]
accuracy_64 = accuracy_64[accuracy['sampling_rate'] == 1]
accuracy_32 = accuracy[accuracy['sequence_length'] == 32]
accuracy_32 = accuracy_32[accuracy['sampling_rate'] == 1]

show_charts(accuracy_256, 256)
show_charts(accuracy_128, 128)
show_charts(accuracy_64, 64)
show_charts(accuracy_32, 32)

accuracy_64_2 = accuracy[accuracy['sequence_length'] == 64]
accuracy_64_2 = accuracy_64_2[accuracy['sampling_rate'] == 2]
accuracy_128_2 = accuracy[accuracy['sequence_length'] == 128]
accuracy_128_2 = accuracy_128_2[accuracy['sampling_rate'] == 2]
accuracy_256_2 = accuracy[accuracy['sequence_length'] == 256]
accuracy_256_2 = accuracy_256_2[accuracy['sampling_rate'] == 2]

show_charts(accuracy_64_2, 64)
show_charts(accuracy_128_2, 128)
show_charts(accuracy_256_2, 256)

accuracy_256_4 = accuracy[accuracy['sequence_length'] == 256]
accuracy_256_4 = accuracy_256_4[accuracy['sampling_rate'] == 4]

show_charts(accuracy_256_4, 256)
```
