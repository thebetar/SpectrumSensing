```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
accuracy = pd.read_csv('../docs/results/accuracy.csv')
```

```python
def show_charts(chart_df, sequence_length):
    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylim(0, 100)
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title('Accuracy of models with sequence length ' + str(sequence_length))
    plt.show()

    # Find accuracy min
    min_accuracy = chart_df['accuracy'].min()

    # Show accuracy in bar chart with min of 80 (with different colors)
    plt.bar(chart_df['model'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylim(min_accuracy - 2, 100)
    plt.ylabel('Accuracy (%)')
    plt.xlabel('Model')
    plt.title(f'Accuracy of models with sequence length {str(sequence_length)} (min {int(min_accuracy - 2)}%)')
    plt.show()

    # Show training time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['training_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Training time (s)')
    plt.xlabel('Model')
    plt.title('Training time of models with sequence length ' + str(sequence_length))
    plt.show()

    # Show response time in bar chart (with different colors)
    plt.bar(chart_df['model'], chart_df['avg_response_time'], color=['blue', 'red', 'green', 'purple', 'orange'])
    plt.ylabel('Response time (ms)')
    plt.xlabel('Model')
    plt.title('Response time of models with sequence length ' + str(sequence_length))
    plt.show()
```

```python
accuracy_256 = accuracy[accuracy['sequence_length'] == 256]
accuracy_256 = accuracy_256[accuracy['sampling_rate'] == 1]

show_charts(accuracy_256, 256)
```

```python
def show_accuracy_sequence_length(chart_df, model_name):
    # Create new column where each unique sequence length + sample_rate is a label
    chart_df = chart_df.sort_values(by='sequence_length')
    chart_df['chart_label'] = chart_df['sequence_length'].astype(int).astype(str) + ' | ' + chart_df['sampling_rate'].astype(str)

    # Show accuracy in bar chart (with different colors)
    plt.bar(chart_df['chart_label'], chart_df['accuracy'], color=['blue', 'red', 'green', 'purple', 'orange'])

    # Set minimum to min value of accuracy - 2
    plt.ylim(bottom=60, top=100)

    plt.ylabel('Accuracy (%)')
    plt.xlabel('Sequence length (sequence length | sampling rate)')
    plt.xticks(rotation=45)
    plt.title(f'Accuracy of {model_name} model per sequence length')
    plt.show()

accuracy_conv = accuracy[accuracy['model'] == 'conv']
accuracy_conv_svm = accuracy[accuracy['model'] == 'conv_svm']
accuracy_fft = accuracy[accuracy['model'] == 'fft']

show_accuracy_sequence_length(accuracy_conv, 'Convolutional')
show_accuracy_sequence_length(accuracy_conv_svm, 'Convolutional SVM')
show_accuracy_sequence_length(accuracy_fft, 'FFT')

accuracy_conv_256 = accuracy_conv[accuracy_conv['sequence_length'] == 256]
accuracy_conv_svm_256 = accuracy_conv_svm[accuracy_conv_svm['sequence_length'] == 256]
accuracy_fft_256 = accuracy_fft[accuracy_fft['sequence_length'] == 256]

show_accuracy_sequence_length(accuracy_conv_256, 'Convolutional 256 samples')
show_accuracy_sequence_length(accuracy_conv_svm_256, 'Convolutional SVM 256 samples')
show_accuracy_sequence_length(accuracy_fft_256, 'FFT 256 samples')
```
